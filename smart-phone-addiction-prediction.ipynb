```python
train = pd.read_csv('/kaggle/input/competitions/playground-series-s6e8/train.csv')
test = pd.read_csv('/kaggle/input/competitions/playground-series-s6e8/test.csv')
sample_sub = pd.read_csv('/kaggle/input/competitions/playground-series-s6e8/sample_submission.csv')
```

```python
import numpy as np
import pandas as pd
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, accuracy_score

# 1. Feature setup
num_cols = [
    'age', 'daily_screen_time_hours', 'social_media_hours', 
    'gaming_hours', 'work_study_hours', 'sleep_hours', 
    'notifications_per_day', 'app_opens_per_day', 'weekend_screen_time'
]
cat_cols = ['gender', 'stress_level', 'academic_work_impact']
features = num_cols + cat_cols
target = 'addicted_label'

# Convert categorical columns to 'category' dtype
X = train[features].copy()
y = train[target].copy()
X_test = test[features].copy()

for col in cat_cols:
    X[col] = X[col].astype('category')
    X_test[col] = X_test[col].astype('category')

# 2. Stratified 5-Fold Cross-Validation setup
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

oof_lgb = np.zeros(len(train))
oof_xgb = np.zeros(len(train))
test_preds_lgb = np.zeros(len(test))
test_preds_xgb = np.zeros(len(test))

# 3. K-Fold Training Loop
for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    print(f"--- Fold {fold + 1} ---")
    X_tr, y_tr = X.iloc[train_idx], y.iloc[train_idx]
    X_va, y_va = X.iloc[val_idx], y.iloc[val_idx]

    # LightGBM Model
    lgb = LGBMClassifier(
        n_estimators=1000,
        learning_rate=0.05,
        num_leaves=63,
        random_state=42,
        n_jobs=-1,
        verbose=-1
    )
    lgb.fit(X_tr, y_tr)
    oof_lgb[val_idx] = lgb.predict_proba(X_va)[:, 1]
    test_preds_lgb += lgb.predict_proba(X_test)[:, 1] / skf.n_splits

    # XGBoost Model
    xgb = XGBClassifier(
        n_estimators=1000,
        learning_rate=0.05,
        max_depth=6,
        enable_categorical=True,
        tree_method='hist',
        random_state=42,
        n_jobs=-1
    )
    xgb.fit(X_tr, y_tr)
    oof_xgb[val_idx] = xgb.predict_proba(X_va)[:, 1]
    test_preds_xgb += xgb.predict_proba(X_test)[:, 1] / skf.n_splits

# 4. Out-of-Fold Blending
oof_ensemble = 0.5 * oof_lgb + 0.5 * oof_xgb
final_test_preds = 0.5 * test_preds_lgb + 0.5 * test_preds_xgb

# 5. Calculate Metrics
# ROC AUC
lgb_auc = roc_auc_score(y, oof_lgb)
xgb_auc = roc_auc_score(y, oof_xgb)
ensemble_auc = roc_auc_score(y, oof_ensemble)

# Accuracy (using 0.5 threshold on predicted probabilities)
oof_preds_binary = (oof_ensemble >= 0.5).astype(int)
ensemble_acc = accuracy_score(y, oof_preds_binary)

print("\n================ Cross-Validation Results ================")
print(f"LightGBM Out-of-Fold ROC AUC : {lgb_auc:.5f}")
print(f"XGBoost Out-of-Fold ROC AUC  : {xgb_auc:.5f}")
print(f"Ensemble Out-of-Fold ROC AUC : {ensemble_auc:.5f}")
print(f"Ensemble Out-of-Fold Accuracy: {ensemble_acc * 100:.2f}%")
print("==========================================================")

# 6. Save Submission File
submission = sample_sub.copy()
submission['addicted_label'] = final_test_preds
submission.to_csv('submission.csv', index=False)
print("Submission file saved successfully as 'submission.csv'!")
```
